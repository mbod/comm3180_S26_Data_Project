## Federal changes to admissions policies may have unexpected lasting impacts

In 2023 the American Supreme Court ruled against Affirmative Action—defined by the [ACLU as](https://assets.aclu.org/live/uploads/document/affirmative_action99.pdf) a set of proactive measures intended to prevent discrimination and ensure equal opportunities in education for historically excluded groups. Alongside this landmark decision, the Trump administration has been attempting to dismantle race-based policies at universities that target specific racial or socioeconomic groups to increase the equity of opportunity in education. 

While university affirmative action programs are controversial, there is evidence to demonstrate that within classrooms, the kind of racial diversity they are intended to create are beneficial to students. Previous studies have found that educational environments have a range of benefits for students on the classroom basis. [Lau](https://doi.org/10.1016/j.labeco.2022.102254), for example, suggests that "higher degree of classroom diversity increases GPA at graduation, improves first year grades of female students, and affects the major choice of white students." It remains to be seen whether these benefits are also present at the institutional level. In the following analysis, we suggest that *the racial and demographic diversity of a university can increase the median earnings of its graduates?* We assume that greater racial homogeneity will have adverse outcomes for students at primarily white institutions (though not necessarily at HBCUs). We also analyze the diversity of athletic programs, and whether the additional resources afforded to different schools for athletics changes outcomes at the school level.  

The analysis brings light to the way that the effects of diversity may last far beyond a university education. We may tie these measures of success to the ability of students to thrive in workplaces where they will inevitably encounter diversity, increased openness to mobility, exposure to alternative life paths, or even simply the intentionality of a school in determining the support structures for  many students of different backgrounds. 


## Explaining the effects and type of diversity

Of course, this last element is key to whether not diversity, by itself, can explain outcomes at different schools. Where schools are more likely to have students from disadvantaged backgrounds we might expect that in the absence of significant resources, median earnings are likely to be lower on average. Structural disadvantages may be more likely to compound among students who receive financial aid in the form of Pell grants for that reason. 

However, if a school has enough resources—such as at elite universities, state schools with significant government funding, or other specifically designated universities—then schools may be able to counter some of these disadvantages. For this reason, the admission rate—a proxy for how exclusive a university  is, and also how well resourced—is included as a control variable. These ‘controls’ enable analyses to take account of the way that factors that all influence the same outcome may intersect. 

Another such variable is the completion rate of a university. This speaks to the difficulty that student have finishing the required courses, among other factors. Generally, this  helps us account for how well a school supports the success of its undergraduates. This is important when accounting for whether students feel equipped to use their undergraduate education in their careers and beyond. 
We can account for each of these factors in a dataset from the Department of Education, called the [College Scorecard](https://collegescorecard.ed.gov/). We specifically chose  to analyze data from 2022. This allowed us to view data as recent as possible, while capturing the effects of diversity before the Trump administration changed their policies toward university admissions. Additionally, because these results are collected by the US government, the specific demographic groups are measured similarly to the census data, allowing us to understand how the diversity of a university compares to the diversity of surrounding areas. 

This is itself an important factor in the analysis, and we decided to measure diversity in two different ways to account for the ways that it can be measured or conceptualized. The first way is through a method we call "similarity scores" which account for how similar the racial composition of a school is to the racial composition of the United States. On average, the US is a relatively diverse country, with significant racial variation, in particular, compared to similar countries-- 60% of the country identifies as white (non-hispanic) compared to 83% in the UK, for example. 

Finally, median earnings themselves are identified within a for individuals that were federally aided, were working, and were not enrolled in school as of the measurement point. Earnings were measured based on wages and deferred compensation reported via IRS form W-2 plus positive self-employment earnings. These measures were calculated for 1,942 primarily four-year institutions, excluding universities in Puerto Rico, Guam, and the Virgin Islands.

## How might minority-serving institutions be different? 

There are, however, quite a few dots that deserve explanation at the top end of the graph. Certain universities are intentionally designated to serve particular minorities, for example, [Hispanic-Serving Institutions (HSIs)](https://www.ed.gov/about/ed-offices/ope/hispanic-serving-institutions-division--home-page) receive special grant funding from the DOE to assist with strengthening institutional programs, facilities, and services to expand the educational opportunities for Hispanic Americans and other underrepresented populations. Furthermore, the DoE supports 107 Historically Black Colleges and Universities (HBCUs) through targeted grant programs, capital financing, and the White House Initiative on HBCUs, which works to advance educational equity and increase federal funding access. 

For these universities, where racial imbalances may intentionally attempt to address inequities experienced by certain groups, we might expect that outcomes are different, and perhaps better than their counterparts that do not receive special funding. 

![](images/minority_race_props.png){width=700}


The second set of regression results, from designated universities, indicate that the relationship is different for expected earnings at universities reporting higher proportions of White and Asian Students, in particular. This is somewhat counterintuitive-- we might expect that a lack of diversity from non-designated would be less likely to be a predictor of median earnings at designated schools. Therefore it's strange that as the population of White or Asian, students increases, the expected earnings of a school do too. We might explain this in two ways. The first is that the ranking of a school determines its diversity, as we noticed earlier in the graph, higher-earning schools tended to have similar and highly-representative student bodies (with populations closely mirroring that of the US population). 

The second is that better performing universities tend to attract more students of any race, while schools with lower median postgrad earnings that are designated as accommodating certain racial groups are more likely to attract higher proportions of those groups. Importantly, however, the proportions of White (though not Asian) students at designated universities remain lower on average than in non-designated universities and indeed than those groups' proportions in the general population (25% White in the sample vs. 57.2% in the population), meaning that a school does not have to be demographically representative of the population in order to benefit from diversity. 

Importantly, these numbers are not perscriptive, but descriptive. That is, schools do not necessarily increase their median earnings by increasing or decreasing the number of students at their school in a particular racial group. Additionally, the analysis does not account for  other variables that might affect the relationship between structural aspects of income inequality between universities, such as the number of students who originate from low income backgrounds, the exclusivity of the university, and the completion rate of a university. 



# How diversity and demographic similarity to the US affects the relationship between diversity and earnings

Up to this point, we've established that the racial composition of a university--which groups are present, and in what proportions-- has a measurable relationship with what graduates can expect to earn. But composition alone is a blunt instrument. A school where 80% of students are Hispanic tells us something different depending on whether it's located in a predominantly Hispanic county in South Texas or rural Vermont. 

Furthermore, the similarity of a university to the country as a whole may not fully capture the 'diversity' it holds. Historically Black Colleges and Universities (HBCUs), for example, are not by definition racially diverse, although they do represent more members of a racial minority. These colleges, as well as Hispanic Serving Institutions (HSIs) are specifically designated by the department of education as a means of addressing the inequality that Hispanic and Black individuals have experienced in the US. Therefore we also measure the diversity of each state, and the county where the school is located. If a school is located in a predominantly Black or Hispanic region, therefore, we might expect there to be a greater number of students who are Black or of Hispanic origin. The different types of similarity scores account for these regional ties. 

To account for this, we introduced two additional ways of thinking about diversity: how similar a university's student body is to the surrounding population, and how concentrated its student body is overall. 

The first of these measures of diversity, what we call "similarity scores" measure how closely a school's racial makeup mirrors the country, its' state, or county. A score of 1 means that the school is a perfect demographic mirror of the region, while a score closer to 0 means that it looks very different. The HHI 

Additionally, we measure diversity with the Herfindahl-Hirschman Index (HHI), a  measure usually used to determine competitiveness in business, which is calculated by summing the squares of the percentage market shares of all firms in an industry. In this case, however, we used the sum of the squares of the percentage of different racial groups, so schools with significant numbers of different racial groups will have a higher HHI. Conversely, schools that are mostly composed of one racial group-- like HBCU's or Jewish Universities-- will have a lower HHI. 

------------------------------------------------------

Running these measures on their own, without controlling for other factors, the results are surprising. Higher HHIs are associated with lower median earnings, which broadly supports the case for diversity. Similarity scores, show that institutions with demographics more aligned with the national average tend to see higher graduate earnings, while the similarity to a county or state wasn't statistically significant, meaning that it's unlikely to have a bearing on the median earnings of a school.

Just like in the institution-level results, however, adding control variables is important to interpreting these relationships. When we account for the share of students recieving Pell Grants, how selective the university is, and whether it's speicifally designated as an HBCU or HSI, the results are somewhat clearer. This matters since a school might look diverse on paper, but primarily because it serves a predominantly low-income student body. This would independently lower median earnings, while "controlling" for these variables means that we can see the effect of diversity on earnings as though all schools were of an equal income. Separating this and the other factors is what allows us to analyze whether diversity itself has an effect. 

![](images/diversity_partial_regression.png){}

With controls in place, national similarity becomes a positive predictor of earnings. Schools whose student bodies more closely reflect the racial composition of the United States as a whole tend to produce higher-earning graduates, even after accounting for income, background, and selectivity. County-level similarity also becomes significant, although the effect is smaller. State-level similarity, however, is is insignificant througout. Therefore while a school's similarity to the country or its very local community matters, the middle ground seems to have less of an effect on earnings. The plots below illustrate this relationship in more detail. 

![](images/similarity_scatter2.png){}

Just as earlier, the relationship is not primarily about any one racial group being "better" for earnings. The highest-earning cshools in the dataset tend neither to be the whitest nor the most homogenously minority; they sit somewhere int he middle, especially with student bodies that reflect the country's demographics. If we account for whether a university is specifically racially designated, or has existing measurs of inequity, particularly with income, the type of diversity becomes more important. 

![](images/ols_coefficients.png){}

This graph demonstrates that socioeconomic background of a university may in fact may be more important than the diversity of a university-- confirming that socioeconomic status, especially as it it intersects with race, is an important factor in the outcomes that students may expect from attending different institutions. These results have some negative and some positive implications. 

On the negative side: pell grant percentages, diversity indices, and admissions rates reduce expected earnings in this analysis. The admissions rate finding is intuitive; more selective schools produce higher earners. The diversity index finding confirms what earlier analyses told us, which is that *without* controls, it appears that diversity hurts outcomes, while compressing those results y adding Pell controls suggests that the effect is truly poverty in disguise. 

On the positive side, national similarity, HBCU status, and county similarity increase expected earnings. The HBCU result implies that once you strip out socioeconomic disadvantage, HBCUs outperform expectations. This is a different story than the earlier analysis, which without this nuance, seemed to imply that taking diversity into account intentionally had adverse outcomes. This outcome is directly relevant to the policy changes being made by the national government, and implies that the changes to affirmative action policies as a means of creating race-blind university systems may not be the best way to deal with structural inequalities. However the non-results of HSIs and PBIs mean that not all racially designated universities benefit from these structures, even where federal funding is present. It might be that that the longevity and well known status of HBCUs contributes to the comparative success of these institutions. 

Across these analyses, however, it's important to note that the statistical measure of the importance of diversity only account for about 25% of the results we see. Diversity is one of many factors involved in determining the expected earnings at a unviversity.  



# Athletic Division Correlations to Diversity 

When we set out to examine the relationship between NCAA division level and racial demographics, we expected to find some variation. The pattern that emerged was greater than anticipated and gives strong insights into diffferent elements of diverstiy leading to success.

First, we started by looking at the correlation between average median earnings and division level. Even though the median earnings data reflects the entire student body rather than student athletes alone, it still gives a meaningful signal about the overall institutional environment that athletes are graduating from.

![Average Median Earnings & Atheltic Divsion](images/division.diversity.png){width=500}


The median earnings boxplot provided above shows that NCAA Division I schools produce graduates with noticeably higher median earnings than both Division II and NAIA institutions, with a median hovering around $67,000 compared to roughly $57,000 and $54,000 respectively. This was counter to our original hypothesis, which predicted that DII and NAIA schools might outperform DI by keeping students more academically focused. 

-------------------------

This led us to then look at possible reasons for what could be leading to these results. Already knowing there are strong correlations between high diversity and sucess in education, we looked at the diverstiy makeup of each NCAA division level.

![Diverstiy Spread Of NCAA Divisions](images/median.division.png){width=500}

What this data reveals is a clear racial difference between diverstiy levels in Division 1 schools versus Division 3 schools. Division I schools carry the highest share of Black students at around 23%, and that proportion drops steadily as you move down to Division II and Division III. 

White student representation moves in exactly the opposite direction, growing larger as division level decreases. Division 3 schools hold the lowest racial diverstiy, and the highest white demographic. In contrast, Division 1 schools overall have the highest diverstiy in the NCAA.

The most likely explanation ties directly back to how each division operates. Division I programs are high-profile, heavily resourced, and recruit student athletes with a deep resource pool. Division III schools, by contrast, offer no athletic scholarships at all. They have limited funds to recruit athletes, and therefore have less of a picking of who commits to their schools. 

Without that financial access point, the demographics of who can realistically attend skews significantly whiter. This suggests that scholarship access and recruitment structure aren't just athletic decisions that soley shape the ahtletic community. They make real impacts on shaping the sucess and diverstiy of entire college campuses.



# What does this analysis tell us about the government's new affirmative action policies? 

Whether or not someone agrees that affirmative action is the best way to ensure diversity, American universities have a clear track record; increasing both resources and diversity has a positive association with increasing median earnings. Although the connections between elite schools' admission processes and the outcomes they create are likely more complex than regression can unvcover in a single analysis, the results of the analysis imply that recent changes to race conscious admissions policies may have adverse affects. However the mixed results form racially designated universities imply that increasing diversity alne may not increase outcomes for students, and instead institutional support relies on the longevity and name-brand recognition of a school. That is, ensuring that students from different backgrounds benefit from America's broad range of perspectives also requires acknowledging class differences and racial biases that determine students' ability to attend and thrive in universities.
